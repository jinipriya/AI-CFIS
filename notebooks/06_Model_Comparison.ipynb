# 06 - Model Comparison

## Objective
The objective of this notebook is to compare the performance of two sentiment analysis models:

1. Logistic Regression with TF-IDF
2. DistilBERT Transformer Model

Both models are evaluated using:
- Accuracy
- Precision
- Recall
- F1-score

The goal is to identify the better-performing model for the Customer Feedback Intelligence System.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
results = {
    "Model": ["Logistic Regression", "DistilBERT"],
    "Accuracy": [0.8885, 0.9325],
    "Precision": [0.89, 0.9329],
    "Recall": [0.89, 0.9319],
    "F1 Score": [0.89, 0.9324]
}

comparison_df = pd.DataFrame(results)

comparison_df
```

```python
comparison_df.set_index("Model").plot(
    kind="bar",
    figsize=(10, 6),
    ylim=(0, 1),
    width=0.7
)

plt.title("Logistic Regression vs DistilBERT")
plt.ylabel("Score")
plt.xlabel("Model")
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Conclusion

Based on the evaluation results, DistilBERT outperformed
Logistic Regression across all four evaluation metrics.

DistilBERT achieved an accuracy of 93.25%, compared to
88.85% for Logistic Regression.

Therefore, DistilBERT is selected as the primary sentiment
analysis model for the AI-Based Customer Feedback
Intelligence System.

Logistic Regression is retained as a lightweight baseline
model for comparison.
